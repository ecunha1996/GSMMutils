```python
new_lipids = ["dgcc140160",
              "dgcc140161",
              "dgcc140183",
              "dgcc140183y",
              "dgcc140184",
              "dgcc140205",
              "dgcc140226",
              "dgcc160160",
              "dgcc160161",
              "dgcc160181",
              "dgcc160181y",
              "dgcc160182",
              "dgcc160183",
              "dgcc160183y",
              "dgcc160184",
              "dgcc160204",
              "dgcc160205",
              "dgcc160226",
              "dgcc161161",
              "dgcc161205",
              "dgcc161226",
              "dgcc162205",
              "dgcc180205",
              "dgcc181181",
              "dgcc181y181y",
              "dgcc181205",
              "dgcc181y205",
              "dgcc181226",
              "dgcc181y226",
              "dgcc182205",
              "dgcc183205",
              "dgcc183y205",
              "dgcc184184",
              "dgcc184205",
              "dgcc184225",
              "dgcc184226",
              "dgcc202205",
              "dgcc203205",
              "dgcc204205",
              "dgcc205205",
              "dgcc205226",
              "dgcc226226",
              "dgcc140140"]

chains_map = {"140": "tetradecanoyl", "160": "hexadecanoyl", "161": "(9Z-hexadecenoyl)",  "162": "(9Z,12Z-hexadecadienoyl)",
              "180": "octadecanoyl", "181": "(9Z-octadecenoyl)", "181y": "(11Z-octadecenoyl)",
              "182": "(9Z,12Z-octadecadienoyl)", "183": "(9Z,12Z,15Z-octadecatrienoyl)", "183y": "(6Z,9Z,12Z-octadecatrienoyl)", "184": "(6Z,9Z,12Z,15Z-octadecatetraenoyl)",
              "202": "(11Z,14Z-eicosadienoyl)", "203": "(8Z,11Z,14Z-eicosatrienoyl)", "204": "(5Z,8Z,11Z,14Z-eicosatetraenoyl)", "205": "(5Z,8Z,11Z,14Z,17Z-eicosapentaenoyl)",
              "225": "(7Z,10Z,13Z,16Z,19Z-docosapentaenoyl)", "226": "(4Z,7Z,10Z,13Z,16Z,19Z-docosahexaenoyl)"}

formulas_map = {
    "140": "C14H27O2",
    "160": "C16H31O2",
    "161": "C16H29O2",
    "162": "C16H27O2",
    "180": "C18H35O2",
    "181": "C18H33O2",
    "181y": "C18H31O2",
    "182": "C18H31O2",
    "183": "C18H29O2",
    "183y": "C18H29O2",
    "184": "C18H27O2",
    "202": "C20H35O2",
    "203": "C20H33O2",
    "204": "C20H31O2",
    "205": "C20H29O2",
    "225": "C22H33O2",
    "226": "C22H31O2"
}
head_formula = "C10H19NO3"
```

```python
result = {}
for lipid in new_lipids:
    if lipid[7] == "y":
        chain_1 = chains_map[lipid[4:8]]
        chain_1_abb = lipid[4:8]
        chain_2 = chains_map[lipid[8:]]
        chain_2_abb = lipid[8:]
    else:
        chain_1 = chains_map[lipid[4:7]]
        chain_1_abb = lipid[4:7]
        chain_2 = chains_map[lipid[7:]]
        chain_2_abb = lipid[7:]
        head_group = "3-O-carboxy-(hydroxymethyl)-choline"
        if chain_1 == chain_2:
            if chain_1.startswith("("):
                full_lipid = f"1,2-di-{chain_1}-{head_group}"
            else:
                full_lipid = f"1,2-di{chain_1}-{head_group}"
        else:
            full_lipid = f"1-{chain_1}-2-{chain_2}-{head_group}"
    full_formula = f"{formulas_map[chain_1_abb]} + {formulas_map[chain_2_abb]} + {head_formula}"
    # get correct formula by summing all C, H, N, and O atoms
    formula = {"C": 0, "H": 0, "N": 0, "O": 0}
    for part in full_formula.split(" + "):
           for atom in formula.keys():
                number = re.search(f"{atom}(\d+)", part)
                if number:
                    formula[atom] += int(number.group(1))
    formula_as_str = ""
    for atom, number in formula.items():
        if number > 0:
            formula_as_str += f"{atom}{number}"
    result[lipid] = {"name": full_lipid, "formula": formula_as_str}
    print(result[lipid])
```

```python
as_df = pd.DataFrame(result).T
as_df['external_identifier']= as_df.index
as_df.to_csv(r"C:\Users\Bisbii\PythonProjects\GSMMutils\data\fame2biomass\dgcc.csv", index=False)
```
